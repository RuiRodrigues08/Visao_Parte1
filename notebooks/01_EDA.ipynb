```python
# ==========================================
# =        Configuração do caminho         =
# ==========================================

import sys
import os
import matplotlib.pyplot as plt

# Adiciona a pasta raiz do projeto aos caminhos do Python para conseguir ler a pasta 'src'
sys.path.append(os.path.abspath('..'))

from src.dataset import get_data_splits

# ==========================================
# =   Criar os splits e o inventário       =
# ==========================================

TRAIN_DIR = "../data/GTSRB/Training"

train_df, val_df = get_data_splits(TRAIN_DIR)

print(f"Total de imagens para treino: {len(train_df)}")
print(f"Total de imagens para validação: {len(val_df)}")

# ==========================================
# =         Gráfico de distribuíção        =
# ==========================================

class_counts = train_df["class_id"].value_counts().sort_index()
plt.figure(figsize=(15, 5))
plt.bar(class_counts.index, class_counts.values, edgecolor='black')
plt.title("Distribuição de Classes no Dataset GTSRB (Treino Original)")
plt.xlabel("ID da Classe")
plt.ylabel("Quantidade de Imagens")
plt.xticks(range(0, 43))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
